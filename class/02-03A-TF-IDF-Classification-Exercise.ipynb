## Exercise: Classifying Newsgroup Posts with TF-IDF

Build a classifier that decides whether a Usenet post comes from a newsgroup about *cars* or about *motorcycles*, using TF-IDF features and logistic regression.

The [20 Newsgroups](https://scikit-learn.org/stable/datasets/real_world.html#the-20-newsgroups-text-dataset) dataset (about 18,000 posts in 20 groups) is publicly an you can load it via `scikit-learn`. We use two groups: `rec.autos` and `rec.motorcycles`. Our goal is to distinguish between these two groups based on their content. They share a lot of vocabulary (engines, brakes, riding, driving). The metadata has been stripped, so the only information available is the text of the posts.

## Loading the Data

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import fetch_20newsgroups
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, ConfusionMatrixDisplay

np.random.seed(1223)
```

```python
categories = ["rec.autos", "rec.motorcycles"]

load_args = dict(
    categories=categories,
    remove=("headers", "footers", "quotes"),
    random_state=43,
)

train = fetch_20newsgroups(subset="train", **load_args)
test = fetch_20newsgroups(subset="test", **load_args)

print(f"Training posts: {len(train.data)}, test posts: {len(test.data)}")
print("Classes:", train.target_names)
```

```python
# Class balance in the training set
pd.Series(train.target_names)[train.target].value_counts()
```
